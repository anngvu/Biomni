# Synapse Analysis Tutorial

This tutorial demonstrates Biomni doing analysis with data from Synapse using Biomni's new Synapse integration.

## Setup

Note: To return the most data from search, including private/collaboration data, set `SYNAPSE_AUTH_TOKEN`.

```python
import os
import sys

# Check for token
if os.environ.get("SYNAPSE_AUTH_TOKEN"):
    print("✓ SYNAPSE_AUTH_TOKEN is set")
else:
    print("⚠️  SYNAPSE_AUTH_TOKEN not found. Set it to download data.")
    print(
        "   You can search public data without authentication but setting token is recommended for more data and downloads."
    )
```

## Data reuse, starting from search and some questions

```python
from biomni.agent import A1

agent1 = A1(path="./biomni_data", llm="claude-sonnet-4-20250514")
```

```python
log1 = agent1.go("""Search for Synapse cutaneous neurofibromas datasets. 
               Some questions of interest for me are: 
               1. does treatment work, and if so can you summarize effectiveness? 
               2. is there therapeutic heterogeinity based on patient characteristic, tumor site, administration method, or anything else present in the data?
               3. (most general question) what else seems interesting in the drug treatment data?    
               If there are any datasets that may answer ONE of these questions, see what can be gleaned from the data through a quick analysis. Carefully select one of the datasets for download, avoiding raw data that would require extensive processing.""")
```
